```python
# 함수 생성 -> 매개변수, 함수의 결과에 대한 힌트를 포함
def func_2(n : int) -> int :
    # 1부터 시작해서 자기 자신까지 1씩 증가시키면서 누적 곱
    result = 1
    for i in range(2, n+1, 1):
        result *= i
    return result
```

### 재귀 함수
- 어떤 이벤트에서 자기 자신을 포함하고 다시 자기 자신을 사용하여 정의되는 것을 의미
- 재귀를 효과적으로 사용하면 프로그램을 간결하고 효율성이 좋게 작성이 가능

```python
def func_4(n : int) -> int:
    if n > 0:
        return n * func_4(n-1)
    else:
        return 1
```

```python
# list에 내장된 reverse() 함수를 생성
def list_reverse(_x):
    # 가장 마지막의 데이터를 추출하여 새로운 리스트의 추가
    result = []

    for i in range(1, len(_x) + 1):
        result.append(
            _x[-i]
        )
    return result
```

```python
test_list = [1,2,3,4,5]
```

```python
list_reverse(test_list)
```

```python
def list_reverse2(_x):
    # _x의 복사본 생성
    result = _x.copy()
    for i in range(len(result)//2):
        result[i], result[-1-i] = result[-1-i], result[i]
    return result
```

#### 연습
- 소수를 구하는 함수를 생성하시오
    - 매개변수 : 2개 (_start (기본값 2), _end(기본값 100))
    - 소수란?
        - 1과 자기 자신 이외에는 나누어 떨어지지 않는 수를 의미
            - 조건 : 2부터 시작해서(자기 자신 -1)까지 어떤 정수로도 나누어 떨어지지 않는다.
    - _start부터 _end까지의 수를 반복하여 소수인 데이터를 비어있는 리스트에 추가한다.
    - 소수의 리스트와 반복 횟수를 되돌려준다.

```python
def point(_start = 2, _end = 100) :
    result = []
    cnt = 0
    for i in range(_start, _end+1):
        for j in range(2, i):
            cnt += 1
            if i % j == 0 :
                break
        else:
            result.append(i)
    return result, cnt
```

```python
point(2, 100)
```

```python
# 함수 생성 매개변수(_start, _end) 2개 기본값 설정
def prime_list(_start = 2, _end = 100):
    # 반복횟수를 1씩 증가하기 위한 초기 값 0 설정
    cnt = 0
    # 소수가 들어갈 비어있는 리스트를 생성
    result = []
    # 반복문을 이용하여 _start부터 _end + 1까지 반복
    for n in range(_start, _end + 1):
        # 2부터 자기 자신의 -1 까지 반복을 하는 반복문 생성
        for i in range(2, n):
            # 반복횟수를 1씩 증가
            cnt += 1
            # 나눠서 나머지가 0인가 체크 (조건문)
            if n % i == 0:
                # 참이라면 반복문을 종료
                break
        # 조건문이 거짓이라면
        else :
            # 소수 리스트에 추가
            result.append(n)
    # 소수의 리스트와 반복횟수를 되돌려준다.
    return result, cnt
```

```python
# 함수 생성 매개변수(_start, _end) 2개 기본값 설정
# 개선 : 소수는 2를 제외한 짝수는 포함 x
def prime_list(_start = 2, _end = 100):
    # 반복횟수를 1씩 증가하기 위한 초기 값 0 설정
    cnt = 0
    # 소수가 들어갈 비어있는 리스트를 생성
    # _start가 2라면 result는 -> [2]
    if _start == 2:
        result = [2]
    else:
        result = []
    # 반복문을 이용하여 _start부터 _end + 1까지 반복
    # 짝수는 2를 제외한 나머지는 소수가 될 수 없으므로 홀수만 반복 실행
    # _start가 짝수라면 _start + 1
    if _start % 2 == 0 :
        _start += 1
    for n in range(_start, _end + 1, 2):
        # 2부터 자기 자신의 -1 까지 반복을 하는 반복문 생성
        for i in range(2, n):
            # 반복횟수를 1씩 증가
            cnt += 1
            # 나눠서 나머지가 0인가 체크 (조건문)
            if n % i == 0:
                # 참이라면 반복문을 종료
                break
        # 조건문이 거짓이라면
        else :
            # 소수 리스트에 추가
            result.append(n)
    # 소수의 리스트와 반복횟수를 되돌려준다.
    return result, cnt
```
